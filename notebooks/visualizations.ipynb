```python
import pandas as pd
import matplotlib.pyplot as plt

# Load CSV file
df = pd.read_csv("bad_channels_overview_final.csv")

# Split and flatten the list of removed channels
channels = df['RemovedChannels'].dropna().str.split(',')
all_channels = [ch.strip() for sublist in channels for ch in sublist]

# Count occurrences
channel_counts = pd.Series(all_channels).value_counts().sort_values(ascending=False)

# --- Plot ---
plt.figure(figsize=(10, 5))
bars = plt.bar(channel_counts.index, channel_counts.values, color='skyblue')  # change color here

plt.title("Frequency of Removed Channels Across Subjects")
plt.xlabel("Channel")
plt.ylabel("Number of Subjects Removed From")

# Rotate labels vertically
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
percent_removed = (channel_counts / len(df)) * 100

plt.figure(figsize=(10, 5))
plt.bar(percent_removed.index, percent_removed.values, color='skyblue')  # different color here
plt.title("Percentage of Subjects Each Channel Was Removed From")
plt.xlabel("Channel")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Split the comma-separated lists and count how many channels per subject
df['n_removed'] = df['RemovedChannels'].dropna().apply(lambda x: len([ch.strip() for ch in x.split(',')]))

# Calculate mean and std
mean_removed = df['n_removed'].mean()
std_removed = df['n_removed'].std()

print(f"Average number of removed channels per subject: {mean_removed:.2f} ± {std_removed:.2f}")
```
